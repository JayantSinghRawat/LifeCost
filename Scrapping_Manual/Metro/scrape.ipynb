```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait, Select
from selenium.webdriver.support import expected_conditions as EC
from time import sleep
import json

browser = webdriver.Firefox()
wait = WebDriverWait(browser, 10)

browser.get("https://bhopal.metroroute.co.in/")
sleep(2)

source_select = Select(browser.find_element(By.ID, "sourceStation"))
dest_select = Select(browser.find_element(By.ID, "destinationStation"))

stations = [opt.get_attribute("value") for opt in source_select.options if opt.get_attribute("value")]

results = []

for src in stations:
    for dst in stations:
        if src == dst:
            continue

        source_select.select_by_value(src)
        dest_select.select_by_value(dst)

        browser.find_element(By.ID, "findRouteBtn").click()

        wait.until(EC.visibility_of_element_located((By.ID, "result")))
        sleep(1)

        summary = browser.find_element(By.ID, "journey-summary")
        items = summary.find_elements(By.CLASS_NAME, "detail-item")

        data = {}
        for item in items:
            label = item.find_element(By.TAG_NAME, "strong").text.replace(":", "")
            value = item.find_element(By.TAG_NAME, "span").text
            data[label] = value

        instructions = browser.find_elements(By.CSS_SELECTOR, "#instructions-list li")
        steps = [i.text for i in instructions]

        data["Route Steps"] = steps
        results.append(data)

        print(f"✔ {data['From']} → {data['To']} scraped")

# Save to JSON
with open("bhopal_metro_fares.json", "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

browser.quit()
print("🎉 All routes saved to bhopal_metro_fares.json")
```

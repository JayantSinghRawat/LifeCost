```python
import pandas as pd

path = "/Users/jayant/Desktop/LifeCost/Scrapping_Manual/cpi/Centre group index   Official website of Labour Bureau, Ministry of Labour and Employment,Government of India.xlsx"

df = pd.read_excel(path)

# Fix column names
df.columns = ["SNo", "BaseYear", "State", "Centre", "Group", "Year", "Month", "Index"]

# Drop header row that got imported as data
df = df[df["BaseYear"] != "Base Year"]

# Convert types
df["Year"] = df["Year"].astype(int)
df["Index"] = pd.to_numeric(df["Index"], errors="coerce")

df.head()
```

```python
bhopal = df[
    (df["Centre"].str.upper() == "bhopal") &
    (df["Group"] == "All Items")
].sort_values(["Year", "Month"])

bhopal.head()
```

```python
bhopal["Inflation_YoY_%"] = (
    bhopal["Index"].pct_change(12) * 100
).round(2)

bhopal.tail()
```

```python
bhopal.to_excel("/mnt/data/bhopal_cpi_clean.xlsx", index=False)
bhopal.to_csv("/mnt/data/bhopal_cpi_clean.csv", index=False)
```
